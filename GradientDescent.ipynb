Implement Gradient Descent Algorithm to find the local minima of a function.
For example, find the local minima of the function y=(x+3)² starting from the point x=2.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def f(x):
  return (x+3)**2

def df(x):
  return 2*(x+3)
```

```python
x = 2
learning_rate = 0.2
iterations = 20
```

```python
for i in range(iterations):
  x = x - learning_rate * df(x)
  print(x)
```

```python

x_values = np.linspace(-6, 3, 100)
y_values = (x_values + 3)**2

plt.figure(figsize=(7, 4))

plt.plot(x_values, y_values, linewidth=2, label='y = (x + 3)²')
plt.scatter(-3, 0, color='red', s=100, label='Minimum Point (-3, 0)')

plt.title("Gradient Descent - Finding Minimum")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True, alpha=0.3)
plt.legend()

plt.show()
```
